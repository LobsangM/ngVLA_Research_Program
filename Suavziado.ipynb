### PARA Z=0.5


Queremos simular cómo se vería una galaxia a un redshift de **z = 0.5**, manteniendo un tamaño físico de **30 kpc × 30 kpc**, y partimos de una imagen original de **2900 px × 2900 px**.

### Escala física original (kpc/pix)

Si consideramos que la galaxia ocupa 30 kpc de lado y está muestreada en 2900 píxeles, entonces:

$$
\text{Escala física por píxel} = \frac{30 \, \text{kpc}}{2900 \, \text{pix}} \approx 0.01034 \, \text{kpc/pix}
$$

### Escala cosmológica para z = 0.5

Usando la calculadora cosmológica de Ned Wright ([link](https://www.astro.ucla.edu/~wright/CosmoCalc.html)), obtenemos para z = 0.5:

$$
1 \, \text{arcsec} = 6 \, \text{kpc}
$$

### Tamaño angular aparente de la galaxia

$$
30 \, \text{kpc} \div 6 \, \text{kpc/arcsec} = 5 \, \text{arcsec}
$$

### Escala angular por píxel (arcsec/pix)

$$
\frac{5 \, \text{arcsec}}{2900 \, \text{pix}} \approx 0.001724 \, \text{arcsec/pix}
$$

### Escala en grados por píxel (para el header FITS)

$$
0.001724 \, \text{arcsec/pix} \div 3600 \approx 4.79 \times 10^{-7} \, \text{deg/pix}
$$


# Paso 2 : Modificar el header

```python
from astropy.io import fits

#Para z=0.5

# Imagen original
archivo_original = "ngc6946Ha_I_Ha_ksb2004.fits" 

# Abrimos el archivo FITS
hdul = fits.open(archivo_original)
data = hdul[0].data
header = hdul[0].header.copy()  # hacemos una copia


for key in ['RA', 'DEC']:
    if key in header:
        del header[key]

# Escala angular simulando z=0.5
arcsec_per_pix = 0.001724
deg_per_pix = arcsec_per_pix / 3600

header['CDELT1'] = -deg_per_pix
header['CDELT2'] = deg_per_pix
header['COMMENT'] = "Header modificado para simular z = 0.5 (0.001724 arcsec/pix)"

# Guardar
fits.writeto("fitsimages/imagen_0.5arcsec_z05.fits", data, header, overwrite=True)
hdul.close()

print(deg_per_pix)

print("Se creó el archivo 'imagen_z05.fits' sin errores.")
```
