```python
#Prueba de AplPy

import aplpy
import matplotlib.pyplot as plt
from astropy.io import fits
import numpy as np

# Cargar el archivo FITS
fits_file = 'ngc6946Ha_I_Ha_ksb2004.fits'
data = fits.getdata(fits_file)

# Valores máximos y mínimos de los datos
max_value = np.max(data)
min_value = np.min(data)
vmin = np.percentile(data[data > 0], 5)

print(f'Min value: {min_value}, Max value: {max_value}, Vmin: {vmin}')
print(data)

# crear una figura APLpy
fig = aplpy.FITSFigure(fits_file)

# Mostrar la imagen
fig.show_grayscale(stretch='log', vmin= vmin, vmax= max_value)

# Mostrar con otro filtro
fig.show_circles(xw=1, yw=1, radius=1)

# Añadir una barra de color
fig.add_colorbar()

#Añadir un título
fig.set_title('NGC 6946 Ha Image')

# Añadir una escala de barras
fig.add_scalebar(0.1, '0.1 Degrees' , color = 'white')  # 0.1 degrees, adjust as necessary
#fig.scalebar.show(0.1)

#Añadir una etiqueta a la escala de barras
fig.scalebar.set_label('0.1 degrees')

# Mostrar la figura
plt.show()

# Guardar la figura como un archivo PNG
#fig.save('ngc6946_ha_image.png')
```

```python
#Prueba de AplPy

import aplpy
import matplotlib.pyplot as plt
from astropy.io import fits
import numpy as np

# Cargar el archivo FITS
fits_file = 'imagen_z1_smoothed.fits'
data = fits.getdata(fits_file)

# Valores máximos y mínimos de los datos
max_value = np.max(data)
min_value = np.min(data)
vmin = np.percentile(data[data > 0], 5)

print(f'Min value: {min_value}, Max value: {max_value}, Vmin: {vmin}')
print(data.shape)

header = fits.getheader('imagen_z1_smoothed.fits')
print(header)

# crear una figura APLpy
fig = aplpy.FITSFigure(fits_file)

# Mostrar la imagen
fig.show_grayscale(stretch='log', vmin= vmin, vmax= max_value)

# Añadir una barra 
fig.add_scalebar(0.5/3600)
fig.scalebar.set_label('0.5 arcsec')
fig.add_beam(major=0.2/3600, minor=0.2/3600, angle=0, label='Beam size: 0.2 arcsec')


#Añadir un título
fig.set_title('NGC 6946 Ha Image')

#CONTRASTE , REDSHIFTS , 
# 0.05, 0.1, 0.15, 0.2 
# Imregrid (Casa)

# Mostrar la figura
plt.show()
```
