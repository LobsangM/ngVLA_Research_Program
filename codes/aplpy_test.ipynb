```python
#Libraries 

import aplpy 
import numpy as np
import astropy.io.fits as fits
import matplotlib.pyplot as plt
```

```python
# Minimun and maximun 

data = fits.getdata('ngc6946Ha_I_Ha_ksb2004.fits')
vmin = np.nanmin(data)
vmax = np.nanmax(data)
```

To start off, use the FITSFigure class to create a canvas to where your data will be plotted, and display the image using a grayscale stretch:

and display the image using a grayscale stretch:

There are a number of arguments that can be passed to show_contour() to control the appearance of the contours as well as the number of levels to show. For more information, see the see the show_contour() documentation. ([Contour_Documentation](https://aplpy.readthedocs.io/en/stable/api/aplpy.FITSFigure.html))

Display a coordinate grid using:

```python

#Crear el canvas
gc = aplpy.FITSFigure('ngc6946Ha_I_Ha_ksb2004.fits')

# Imagen en escala de Grises
gc.show_grayscale()

#Contornos sobre la imagen en grises
gc.show_contour('ngc6946Ha_I_Ha_ksb2004.fits', colors='yellow')

# Imagen a color
gc.show_colorscale(cmap='nipy_spectral', stretch='log' , vmin=0.01, vmax=vmax)

# Añadir grid
gc.add_grid()

# Quitar Grid
#gc.remove_grid()

#Plotear la imagen
plt.show()
```

A grayscale or colorscale representation of the FITS image can be shown and hidden using the following methods:

```python
fig = aplpy.FITSFigure('ngc6946Ha_I_Ha_ksb2004.fits')

fig.show_grayscale()
#fig.hide_grayscale()
fig.show_colorscale()
#fig.hide_colorscale()
```

The figure can be interactively explored by zooming and panning. To recenter on a specific region programmatically, use the following method, specifying either a radius. or a separate width and height.

To see which values use, we can do this:

```python

from astropy.io import fits
from astropy.wcs import WCS

hdul = fits.open('ngc6946Ha_I_Ha_ksb2004.fits')
wcs = WCS(hdul[0].header)

data = hdul[0].data
ny, nx = data.shape  # cuidado con el orden y,x
x_center = nx / 2
y_center = ny / 2

ra_dec = wcs.pixel_to_world(x_center, y_center)
# print(ra_dec.ra.deg, ra_dec.dec.deg)
```

So, we have now

```python
fig.recenter(ra_dec.ra.deg, ra_dec.dec.deg, radius=0.1)  # degrees

#fig.recenter( ra_dec.ra.deg, ra_dec.dec.deg , width=0.3, height=0.2)  # degrees

plt.show()
```
