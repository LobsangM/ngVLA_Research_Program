```python
from astropy.io import fits
import numpy as np
```

```python
# Configurar Para VLA
fits_file = "VLA_image.fits"  
output_file = "VLA_umbral.fits"  
```

```python
# umbral en microJy/beam
umbral = 2.4   #VLA

#abrir fits
with fits.open(fits_file) as hdul:
    data = hdul[0].data.copy()  
    header = hdul[0].header
```

```python
# --- Aplicar el umbral ---
mask = data < umbral
data[mask] = 0.0

# --- Guardar la nueva imagen ---
hdu = fits.PrimaryHDU(data=data, header=header)
hdu.writeto(output_file, overwrite=True)

print(f"Imagen procesada guardada en: {output_file}")
```
